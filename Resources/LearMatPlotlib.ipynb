```python
tradding_days = [1,2,3,4,5]
original_prices = [10,15,20,25,30]
predicted_prices = [15,15,18,25,28]
```

```python
plt.plot(tradding_days, original_prices, 'b-o', label='Original Price');
plt.plot(tradding_days, predicted_prices, 'r-o', label='Predicted Price');
plt.legend();
plt.xlabel("Tradding Day");
plt.xlabel("Price");
plt.savefig('example_prediction', dpi=300)
```

```python
plt.subplot(3, 1, 1)
plt.plot(tradding_days, original_prices, 'b-o', label='Original Price');
plt.legend();
plt.xlabel("Tradding Day");
plt.xlabel("Price");

plt.subplot(3, 1, 2)
plt.plot(tradding_days, predicted_prices, 'r-o', label='Predicted Price');
plt.legend();
plt.xlabel("Tradding Day");
plt.xlabel("Price");

# price_diff = [o-p for o, p in zip(original_prices, predicted_prices)]
# plt.subplot(3, 1, 3)
# plt.bar(tradding_days, price_diff, color="purple", label='Price');
# plt.legend();
# plt.xlabel("Tradding Day");
# plt.xlabel("Price");

plt.tight_layout();
```
